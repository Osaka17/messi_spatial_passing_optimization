```python
messi_super_clean['change_x'] = messi_clean['end_x'] - messi_clean['start_x']
messi_super_clean['change_y'] = messi_clean['end_y'] - messi_clean['start_y']
```

```python
messi_super_clean.head()
```

```python
plt.plot([60,60], [0,80], color='white')
plt.plot([0, 18, 18, 0], [18, 18, 62, 62], color='white')
plt.plot([120, 102, 102, 120], [18, 18, 62, 62], color='white')
plt.xlim(0, 120)
plt.ylim(80,0)
plt.gca().set_facecolor('#2d5a27')
plt.grid(color='white', linestyle='--', linewidth=0.5, alpha=0.3)

plt.title('Lionel Messi — Passing Vector Map\nvs. Villarreal | La Liga 2014/15', fontsize=12, fontweight='bold', pad=15)
plt.xlabel('Pitch Length (Meters)', fontweight='semibold', fontsize=10)
plt.ylabel('Pitch Width (Meters)', fontweight='semibold', fontsize=10)

plt.xticks(range(0, 121, 10))
plt.yticks(range(0, 81, 10))
plt.tick_params(
    top=True, 
    bottom=True, 
    right=True, 
    left=True, 
    labeltop=True,
    labelbottom=True,
    labelright=True,
    direction='out'
)

plt.quiver(
    messi_super_clean['start_x'], 
    messi_super_clean['start_y'], 
    messi_super_clean['change_x'], 
    messi_super_clean['change_y'],
    color='yellow',
    angles='xy',
    scale_units='xy',
    scale=1,
    width=0.007,
    headwidth=3,
    alpha=0.75,
    edgecolors='white'
)
```

```python
messi_super_clean.head()
```

```python
import numpy as np

messi_super_clean['far_120_x'] = 120 - (messi_super_clean['end_x'])
messi_super_clean['start_far_mainpoint'] = np.sqrt((120 - (messi_super_clean['start_x']))**2 + (40 - (messi_super_clean['start_y']))**2)
messi_super_clean['end_far_mainpoint'] = np.sqrt((120 - (messi_super_clean['end_x']))**2 + (40 - (messi_super_clean['end_y']))**2)

```

```python
messi_super_clean.head()
```

```python
messi_super_clean['absolute_drop'] = messi_super_clean['start_far_mainpoint'] - messi_super_clean['end_far_mainpoint']
```

```python
messi_super_clean['absolute_drop'].head(10)
```

```python
match_events["under_pressure"].head()
```

```python
messi_super_clean['progression_score'] = messi_super_clean['change_x'] + messi_super_clean['absolute_drop']
```

```python
messi_super_clean['progression_score'].head()
```

```python
expected_threat = messi_super_clean.sort_values(by='progression_score', ascending=False)
expected_threat.head()
```

```python
expected_threat = expected_threat[['progression_score', 'absolute_drop', 'change_x']]
```

```python
expected_threat.head(5)
```

```python
expected_threat.tail(5)
```
